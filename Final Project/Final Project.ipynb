```python
import pandas as pd
import os
import numpy as np
from scipy.stats import chi2_contingency
```

```python
brfss_dir = '{}/BRFSS2023.csv'.format(os.getcwd())
brfss_original_df = pd.read_csv(brfss_dir)
```

```python
# Only select columns of interest
brfss_df = brfss_original_df[['addepev3', 'firearm5', '_ment14d', '_age80', 'sexvar', '_racegr3', 'primins1', 'educa', 'income3', 'loadulk2', 'children']]
brfss_dir = '{}/BRFSS_subset.csv'.format(os.getcwd())
brfss_df.to_csv(path_or_buf = brfss_dir, index=False)
```

```python
# Cleaning variables
# ADDPEV3: Change 7 and 9 to missing
brfss_df['addepev3'] = brfss_df['addepev3'].replace({7: np.nan, 9: np.nan})

print(brfss_df['addepev3'].value_counts(dropna = False))

# _AGE80
```

```python
# Objective 1: Prevalence of depression among gun owners
depr_firearms_crosstab = pd.crosstab(brfss_df['addepev3'], brfss_df['firearm5'], margins = True)
print(depr_firearms_crosstab)

# Proportions
depr_firearms_prop = depr_firearms_crosstab = pd.crosstab(brfss_df['addepev3'], brfss_df['firearm5'], normalize = 'columns') * 100
print(depr_firearms_prop)

# P-value overall
chisq_pvalue = chi2_contingency(pd.crosstab(brfss_df['addepev3'], brfss_df['firearm5'])).pvalue
print(chisq_pvalue)
```
